In this notebook, we load the MNIST Data Set and look at some instances of the data.

```python
# libraries
import numpy as np
import matplotlib.pyplot as plt

# mnist data
from sklearn.datasets import fetch_openml

# the dataset: 
mnist = fetch_openml('mnist_784', as_frame = False) # as_frame = False to get numpy arrays

# the data:
X, y = mnist['data'], mnist['target']

print(X.shape) # matrix of 70000 images of 28x28 pixels flattened to 784-dimensional vectors
print(y) # labels as strings representing digits 0-9
```

Theh, we can show a few samples:

```python
# plot some samples
fig, axes = plt.subplots(3, 5, figsize=(10, 6))
for i, ax in enumerate(axes.flat):
    ax.imshow(X[i].reshape(28, 28), cmap='binary')
    ax.set_title(f'Label: {y[i]}')
    ax.axis('off')

# save and show the plot
plt.savefig('../../images/mnist_samples.png', dpi=300)
plt.show()
```
