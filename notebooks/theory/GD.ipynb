Here is a very simple visual example of gradient descent.

```python
# libs
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

We encode the function I use as an example in the notebook:

$$
f(x) = 2x^2 + y^2
$$

and the gradient.

```python
def f(x, y):
    return x**2 + y**2

def grad_f(x, y):
    dfdx = 2*x
    dfdy = 2*y
    return np.array([dfdx, dfdy])
```

Once we have this, we plot this and the gradient in 3D:

```python
# heatmap of f with gradient vectors

# data for heatmap
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# data for gradient at x0, y0
x0, y0 = 2.0, 2.0
grad = - grad_f(x0, y0)
U, V = grad[0], grad[1]

# plot
plt.figure(figsize=(7, 5))
plt.contourf(X, Y, Z, levels=50, cmap='viridis', alpha=0.5)
plt.colorbar(label=r'$f(x, y)$')
plt.quiver(x0, y0, U, V, color='red', scale=50, width=0.005)
plt.scatter(x0, y0, color='black', s=100, label='Point (2,2)')
plt.scatter(0, 0, color='black', s=100, marker='x', label='Minimum (0,0)')
plt.title(r'Heatmap of $f(x, y)$ with Gradient Vector at (2, 2)', family='serif', size=12)
plt.legend()
plt.xlabel(r'$x$')
plt.xlim(-5, 5)
plt.ylabel(r'$y$')
plt.ylim(-5, 5)

# save 
plt.savefig("../../images/gradient_descent_heatmap.png", dpi=300)
plt.show()
```
