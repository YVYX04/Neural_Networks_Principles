This is a very small notebook used to generate a plot of the sigmoid function and integrate it in my summary of Nielsen (2015).

As our code base will be in python, we also encode the sigmoid function in python:

```python
# libraries
import numpy as np
import matplotlib.pyplot as plt
```

Once the libraries are loaded, we encode the function:

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

And we create a plot of this function:

```python
# x values
x = np.linspace(-5, 5, 50)

# y values
y = sigmoid(x)

# plotting
fig, ax = plt.subplots(figsize=(6,3))
ax.plot(x, y, color='black')
ax.scatter(0, 0.5, color='red', zorder=5)  # point at (0, 0.5)

# labels and aesthetics
ax.set_title('Sigmoid Activation Function', family='serif', size=12)
ax.set_xlabel('z', family='serif', size=10)
ax.set_ylabel(r'$\sigma(z)$', family='serif', size=10)
ax.grid(alpha = 0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# save figure
plt.savefig('../../images/sigmoid_function.png', dpi=300)
plt.show()
```
